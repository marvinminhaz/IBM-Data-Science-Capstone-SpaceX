```python
import pandas as pd
import wget
```

```python
wget.download('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_dash.csv', out='../data/raw/')
```

```python
dash_df = pd.read_csv('../data/raw/spacex_launch_dash.csv')
print('File downloaded and read.')
dash_df.head()
```

```python
col_names = dash_df.columns
col_names
```

```python
dash_df = dash_df.rename(columns={
    'Flight Number': 'Flight_Number',
    'Launch Site': 'Launch_Site',
    'Payload Mass (kg)': 'Payload_Mass_kg',
    'Booster Version' : 'Booster_Version',
    'Booster Version Category' : 'Booster_Version_Category'
})
dash_df = dash_df.drop('Unnamed: 0', axis=1)
dash_df.head()
```

```python
launch_sites = list(dash_df.Launch_Site.values.unique())
launch_sites
```

```python
dash_df.Payload_Mass_kg.describe()
```

```python
import dash
import plotly.io as pio
import plotly.graph_objects as go 
import dash_html_components as html
# import dash_core_components as dcc
from dash import Dash, html, dcc
from dash.dependencies import Input, Output
import plotly.express as px
```

```python
app = dash.Dash(__name__)
```

```python
app.layout = html.Div(children=[
    html.H1(
        'SpaceX Launch Records Dashboard',
        style={'text-align': 'center', 'color': '#503D36', 'font-size': '40'}
    ),
    html.Br(),

    dcc.Dropdown(launch_sites, 'ALL', id='site-dropdown'),

    html.Br(),

    html.Div(dcc.Graph(id='success-pie-chart')),

    html.Br(),

    html.P('Payload range (Kg):'),

    dcc.RangeSlider(
        dash_df.Payload_Mass_kg.min(), 
        dash_df.Payload_Mass_kg.max(), 
        value=[2121, 3412.50], id='payload-slider'
    ),

    html.Br(),

    html.Div(dcc.Graph(id='success-payload-scatter-chart'))
])

@app.callback(
    Output(component_id='success-pie-chart', component_property='figure'),
    Input(component_id='site-dropdown', component_property='value')
)
```

```python
pio.renderers.default = 'notebook_connected'
fig=go.Figure()
fig
```

```python
fig = px.pie(dash_df, values='class', names='Launch_Site')
fig.show()
```

```python
launch_site_sucesses = dash_df.groupby('Launch_Site')['class'].value_counts().reset_index()
launch_site_sucesses
```

```python
launch_site = 'CCAFS LC-40'
site_df = dash_df[dash_df['Launch_Site'] == launch_site]
site_df['Outcome'] = site_df['class'].map({0:'Failure',1:'Success'})
site_df
```

```python
fig = px.pie(site_df, names='Outcome', title=f'Success vs Failure for {launch_site}')
fig.show()
```
